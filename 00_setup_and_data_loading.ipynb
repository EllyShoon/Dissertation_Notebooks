```python
# ── Cell 5: Schema validation ────────────────────────────────
# Verify that the key columns expected by the analysis exist
# in each dataset before any further processing.

REQUIRED_MT = [
    'participantcode', 'round', 'sessioncode',
    'liberal', 'political_orientation',
    'trust_scientists', 'environment_worry',
    'environment_responsib', 'environment_natural_p',
    'gender', 'age', 'ethnic_min', 'education', 'prole',
    'extraction', 'extraction_1', 'extraction_2', 'extraction_3',
    'cpr1groupgrp_treat_num'
]

REQUIRED_PR = REQUIRED_MT + [
    'election_2020', 'election_2024',
    'risk_attitude', 'short_long_reward',
    'environment_changing',
    'environment_mov1', 'environment_threat1',
    'environment_beha1', 'environment_uti1', 'environment_growth1',
    'legacy_concern1'
]

def check_schema(df, required, label):
    missing = [c for c in required if c not in df.columns]
    if missing:
        print(f"  ✗ {label}: MISSING columns → {missing}")
    else:
        print(f"  ✓ {label}: all required columns present")

check_schema(df_mt, REQUIRED_MT, 'MTurk')
check_schema(df_pr, REQUIRED_PR, 'Prolific')
```
