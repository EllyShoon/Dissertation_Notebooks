# Notebook 01 — Data Preparation
Covers all data-cleaning, recoding, reverse-scoring, composite construction,
and reliability checks before any inferential analysis.  
All decisions follow **Proposal Section 3.4**, **Milfont et al. (2006)**, and
**Wade-Benzoni & Tost (2009)** as cited in the dissertation.


```python
# ── Cell 1: Load data ────────────────────────────────────
import numpy as np
import pandas as pd
import pingouin as pg
import warnings
warnings.filterwarnings('ignore')

df_mt = pd.read_csv('data/Data_decisions_MA_Mturk.csv')
df_pr = pd.read_csv('data/Data_decisions_MA_Prolific.csv')
print(f"MTurk:   {len(df_mt):,} rows | Prolific: {len(df_pr):,} rows")
```

```python
# ── Cell 2: Treatment condition dummies ──────────────────────

# Four conditions (2×2 factorial):
#   IT&GI    = intergenerational transmission + green incentive
#   IT&NGI   = IT only
#   NIT&GI   = green incentive only
#   NIT&NGI  = pure control  ← reference category (coded 0 on all 3 dummies)
#
# IT_present / GI_present are *collapsed* dummies used ONLY after
# contrasts from 03_replication.ipynb confirm the IT effect is
# positive and significant in both 2×2 columns.

def make_condition_dummies(df):
    c = df['cpr1groupgrp_treat_num']
    df['cond_IT_GI']   = (c.isin(['IT&GI'])).astype(int)
    df['cond_IT_only'] = (c.isin(['IT&NGI', 'IT'])).astype(int)
    df['cond_GI_only'] = (c.isin(['NIT&GI', 'GI'])).astype(int)
    df['cond_control'] = (c.isin(['NIT&NGI'])).astype(int)
    df['IT_present']   = df['cond_IT_GI'] + df['cond_IT_only']
    df['GI_present']   = df['cond_IT_GI'] + df['cond_GI_only']
    return df

df_mt = make_condition_dummies(df_mt)
df_pr = make_condition_dummies(df_pr)

for label, df in [('MTurk', df_mt), ('Prolific', df_pr)]:
    cols = ['cond_IT_GI', 'cond_IT_only', 'cond_GI_only', 'cond_control']
    assert (df[cols].sum(axis=1) == 1).all(), f"{label}: dummy sum != 1"
    print(f"{label}: condition dummies ✓")

print("Condition distribution (Mturk):")
print(df_mt['cpr1groupgrp_treat_num'].value_counts())
print("Condition distribution (Prolific):")
print(df_pr['cpr1groupgrp_treat_num'].value_counts())
```

```python
# ── Cell 3: Reverse-score negatively worded items ────────────
# LEGACY CONCERN (Wade-Benzoni & Tost, 2009):
#   Items 5 & 6 carry negated wording — identified from oTree survey code:
#     lc5: "my actions do NOT have a positive effect on others"
#     lc6: "I feel I have done nothing that will survive after I die"
#   Scale: 1–5 (five_point_scale) → reversed = (5+1) - x = 6 - x
#
# ENVIRONMENTAL CONCERN COMPOSITE (Milfont et al., 2006):
#   Negatively worded items identified from survey item text:
#     mov:    items 2, 3, 5  (don't, NOT, not want)
#     threat: items 4, 5, 6  (too pessimistic, do not believe, wrong)
#     beha:   items 1, 2, 5  (could not, not interested, NOT the kind)
#     uti:    items 1, 5, 6  (jobs > env, secondary, consumer > pollution)
#     growth: items 2, 4, 6  (as many as they wish, never, no right)

LEGACY_REVERSE = ['legacy_concern5', 'legacy_concern6']

ENV_REVERSE = [
    'environment_mov2',    'environment_mov3',    'environment_mov5',
    'environment_threat4', 'environment_threat5', 'environment_threat6',
    'environment_beha1',   'environment_beha2',   'environment_beha5',
    'environment_uti1',    'environment_uti5',     'environment_uti6',
    'environment_growth2', 'environment_growth4', 'environment_growth6',
]

def reverse_score(df, items, scale_max=5):
    """Reversed = (scale_max + 1) − original. Skips absent columns."""
    df = df.copy()
    for col in items:
        if col in df.columns:
            df[col] = (scale_max + 1) - df[col]
    return df

# Prolific has all items; MTurk may have partial env items
df_pr = reverse_score(df_pr, LEGACY_REVERSE + ENV_REVERSE, scale_max=5)
df_mt = reverse_score(df_mt, LEGACY_REVERSE + ENV_REVERSE, scale_max=5)

print("Reverse-scoring complete.")
print(f"  Legacy items reversed:      {LEGACY_REVERSE}")
print(f"  Env. concern items reversed: {len(ENV_REVERSE)} items")
```

```python
# ── Cell 4: Cronbach's Alpha reliability checks ───────────────
# Threshold: α ≥ 0.70 (Milfont et al., 2006; Wade-Benzoni & Tost, 2009).
# If α < 0.70: inspect item-total correlations and consider
# dropping the weakest item or using subscales.

LEGACY_ITEMS = [f'legacy_concern{i}' for i in range(1, 7)]

ENV_ITEMS = (
    [f'environment_mov{i}'    for i in range(1, 7)] +
    [f'environment_threat{i}' for i in range(1, 7)] +
    [f'environment_beha{i}'   for i in range(1, 7)] +
    [f'environment_uti{i}'    for i in range(1, 7)] +
    [f'environment_growth{i}' for i in range(1, 7)]
)

# Standalone items — NOT composited per Proposal Section 3.4:
# "worry, perceived responsibility, and climate change beliefs
#  as discrete, conceptually distinct predictors"
DISCRETE_ENV = [
    'environment_worry',
    'environment_responsib',
    'environment_natural_p',
    'environment_changing',      # Prolific only
]

def cronbach_alpha(df, items, label):
    available = [c for c in items if c in df.columns]
    if len(available) < 3:
        print(f"  ⚠  {label}: fewer than 3 items available — skipping alpha")
        return None
    data = df[available].dropna()
    alpha = pg.cronbach_alpha(data=data)[0]
    status = '✓ Acceptable' if alpha >= 0.70 else '✗ Below threshold — inspect items'
    print(f"  α = {alpha:.3f}  ({len(available)} items)  [{status}]   {label}")
    return alpha

print("=== Prolific ===")
cronbach_alpha(df_pr, LEGACY_ITEMS, 'Legacy Concern')
cronbach_alpha(df_pr, ENV_ITEMS,    'Environmental Concern (full composite)')

# MTurk (Study 1): environmental items are discrete predictors —
# Proposal §3.4 specifies no compositing; Cronbach's α not applicable.
# No reliability check needed or performed for Study 1 env items.
```

```python
# ── Cell 5: Build composite scores ───────────────────────────
# Mean composite preserves the original 1–5 scale and is robust
# to occasional missing items (missing < 20% per participant).

def make_composite(df, items, name):
    available = [c for c in items if c in df.columns]
    df[name] = df[available].mean(axis=1)
    return df

# Prolific only: full composites
df_pr = make_composite(df_pr, LEGACY_ITEMS, 'legacy_concern_comp')
df_pr = make_composite(df_pr, ENV_ITEMS,    'env_concern_comp')

# MTurk: no compositing — env items are discrete predictors (Proposal Section 3.4)
# H4 environmental moderation in Study 1 uses item-level variables directly.

print("Composite descriptives (Prolific):")
print(df_pr[['legacy_concern_comp', 'env_concern_comp']].describe().round(3))
```

```python
# ── Cell 6: Standardise continuous moderators ────────────────
# Z-scoring removes multicollinearity in interaction terms.
# Mean-centred values are also interpretable: β at moderator = mean.

def standardise(df, cols):
    for col in cols:
        if col in df.columns:
            df[f'{col}_z'] = (df[col] - df[col].mean()) / df[col].std()
    return df

MODS = [
    'liberal',
    'legacy_concern_comp',
    'env_concern_comp',
    'environment_worry',
    'environment_responsib',
    'environment_natural_p',
]
df_mt = standardise(df_mt, MODS)
df_pr = standardise(df_pr, MODS)
print("Standardised variables added (suffix _z).")
```

```python
# ── Cell 7: Ideology grouping (both datasets) ─────────────────
# liberal: 1=Extremely liberal … 7=Extremely conservative
#   Liberal      = 1–3
#   Moderate     = 4
#   Conservative = 5–7
# Used descriptively in H1 comparison table.

def add_ideology_group(df):
    df['ideology_group'] = pd.cut(
        df['liberal'],
        bins=[0, 3, 4, 7],
        labels=['Liberal', 'Moderate', 'Conservative']
    )
    return df

df_mt = add_ideology_group(df_mt)
df_pr = add_ideology_group(df_pr)

print("MTurk ideology distribution:")
print(df_mt['ideology_group'].value_counts())


print("\nProlific ideology distribution:")
print(df_pr['ideology_group'].value_counts())
```

```python
# ── Cell 8: Democrat/Republican dummy (Prolific only) ─────────
# PURPOSE: robustness check per Proposal Section 3.5 — replaces the
# continuous ideology scale with a clean partisan binary.
#
# Built from election_2024 (oTree survey: hypothetical vote preference).
# Integer codes from survey source:
#   Democrat  = 1 (Biden), 5 (Buttigieg), 6 (Klobuchar)
#   Republican= 2 (Trump),  3 (Scott),    4 (Haley)
#
# Secondary check via political_orientation (1–7 party ID scale):
#   1=Strong Dem, 2=Dem, 3=Ind near Dem → dem_rep_party_id = 1
#   5=Ind near Rep, 6=Rep, 7=Strong Rep → dem_rep_party_id = 0
#   4=Independent, 8=Other              → NaN (excluded from model)
#
# election_2024 and political_orientation are PROLIFIC-ONLY columns.
# MTurk does not contain either.

def add_dem_rep(df):
    dem_e24  = [1, 5, 6]
    rep_e24  = [2, 3, 4]
    df['dem_rep'] = np.where(
        df['election_2024'].isin(dem_e24), 1,
        np.where(df['election_2024'].isin(rep_e24), 0, np.nan)
    )
    dem_po  = [1, 2, 3]
    rep_po  = [5, 6, 7]
    df['dem_rep_party_id'] = np.where(
        df['political_orientation'].isin(dem_po), 1,
        np.where(df['political_orientation'].isin(rep_po), 0, np.nan)
    )
    return df

df_pr = add_dem_rep(df_pr)

print("dem_rep (election_2024):")
print(df_pr['dem_rep'].value_counts(dropna=False)
      .rename({1.0:'Democrat', 0.0:'Republican'}))
print("\ndem_rep_party_id (political_orientation):")
print(df_pr['dem_rep_party_id'].value_counts(dropna=False)
      .rename({1.0:'Dem-leaning', 0.0:'Rep-leaning'}))
```

```python
# ── Cell 9: Save clean datasets ──────────────────────────────
df_mt.to_csv('data/mturk_clean.csv',    index=False)
df_pr.to_csv('data/prolific_clean.csv', index=False)

print(f"mturk_clean.csv   → {len(df_mt):,} rows, "
      f"{df_mt['participantcode'].nunique()} participants")
print(f"prolific_clean.csv → {len(df_pr):,} rows, "
      f"{df_pr['participantcode'].nunique()} participants")
```
